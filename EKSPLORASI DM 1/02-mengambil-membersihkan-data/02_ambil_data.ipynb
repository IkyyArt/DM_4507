# 02a - Mengambil Data
Materi Modul 2: membaca data dari file CSV/JSON, database, API, dan cloud.

Data yang sama (Wine) tersedia dalam tiga format di folder `datasets/`: `wine.csv`, `wine.json`, `wine.db`.

## CSV

```python
import pandas as pd
import sqlite3

df_csv = pd.read_csv('../datasets/wine.csv')
df_csv.head(3)
```

## JSON

```python
df_json = pd.read_json('../datasets/wine.json')
df_json.head(3)
```

## Database (SQLite)

```python
con = sqlite3.connect('../datasets/wine.db')
df_db = pd.read_sql('SELECT alcohol, proline, target FROM wine WHERE proline > 1000 ORDER BY proline DESC', con)
con.close()
print(df_db.shape)
df_db.head()
```

## Memastikan ketiga sumber identik

```python
print('CSV vs JSON :', df_csv.round(6).equals(df_json.round(6)))
```
