# 03a - Exploratory Data Analysis
Materi Modul 3: EDA dengan visualisasi dan mengelompokkan data.
Dataset: **Wine** (`datasets/wine.csv`, 178 sampel, 13 fitur, 3 kelas).

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style='whitegrid')
```

```python
df = pd.read_csv('../datasets/wine.csv')
df['kelas'] = df['target'].map({0:'class_0', 1:'class_1', 2:'class_2'})
df.head()
```

```python
print(df.shape)
df.info()
```

```python
df.describe().T
```

```python
print('Missing value:', df.isna().sum().sum())
df['kelas'].value_counts()
```

```python
df.drop(columns='target').hist(figsize=(14,10), bins=20)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10,8))
sns.heatmap(df.drop(columns=['target','kelas']).corr(), cmap='coolwarm', center=0)
plt.title('Korelasi antar fitur')
plt.show()
```

```python
sns.boxplot(data=df, x='kelas', y='alcohol')
plt.show()
```

## Mengelompokkan data (groupby)

```python
df.groupby('kelas')[['alcohol','flavanoids','proline']].agg(['mean','std']).round(2)
```

```python
sns.pairplot(df[['alcohol','flavanoids','proline','kelas']], hue='kelas', height=2.2)
plt.show()
```
